```python
# ABOUT THIS NOTEBOOK!
# this notebook is used to extract data in south jakarta bound from overturemaps.

# REQUIREMENT!
# please install those library first before running the code.

# pip install overturemaps
# pip install geopandas
# pip install pandas

# 01 Import the following library.
import pandas as pd
import geopandas as gpd
from overturemaps import core
```

```python
# 02 Import administration boundary in gpkg format and reproject it.
bound = gpd.read_file("../data/admin_boundary.gpkg")
print(bound.crs)

bound = bound.to_crs(epsg=4326)
print(bound.crs)

# 03 Show boundary statements.
bbox = bound.total_bounds
print(bbox)
```

```python
# 04 Extract data.
POI_data = core.geodataframe("place", bbox=tuple(bbox))
road_network_data = core.geodataframe("segment", bbox=tuple(bbox))
building_data = core.geodataframe("building", bbox=tuple(bbox))

print(f'Total POI data : {len(POI_data)} features')
print(f'Total Road Network Data : {len(road_network_data)} features')
print(f'Total Building Footprint Data : {len(building_data)} features')
```
